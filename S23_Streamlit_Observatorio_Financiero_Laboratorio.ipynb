## 01 · Preparar el entorno

**¿Qué es Streamlit?** Una biblioteca de Python que convierte instrucciones como `st.title()` o `st.selectbox()` en componentes de una página web.

| En un notebook | En Streamlit |
|---|---|
| `print(resultado)` muestra texto en una celda | `st.metric()` muestra un indicador |
| `df.plot()` crea un gráfico en el notebook | `st.line_chart()` lo presenta en la aplicación |
| Se cambian variables editando código | El usuario cambia parámetros con controles |

Primero instalamos las bibliotecas. **No ejecutes `streamlit run` directamente con `!` sin segundo plano**, porque la celda quedaría ocupada.

````python

%%writefile app.py

import streamlit as st
import pandas as pd
import numpy as np

st.set_page_config(
    page_title="FinanceLab | Observatorio Financiero",
    page_icon="📈",
    layout="wide"
)

st.markdown("""
<style>
.stApp {
    background-color: #0b1425;
    color: #edf4ff;
}
[data-testid="stSidebar"] {
    background-color: #10243a;
}
[data-testid="stMetric"] {
    background-color: #172b43;
    padding: 15px;
    border-radius: 12px;
    border: 1px solid #29415b;
}
h1, h2, h3 {
    color: #5eead4;
}
</style>
""", unsafe_allow_html=True)

# BANNER
st.image(
    "assets/banner_financiero.png",
    use_container_width=True
)

st.title("📈 FinanceLab")
st.subheader("Mi Observatorio Financiero")
st.caption("Ingeniería Financiera | Laboratorio S23")

# BARRA LATERAL
st.sidebar.header("⚙️ Configuración")

universo = ["AAPL", "MSFT", "NVDA", "JPM", "GOOGL", "SPY", "GLD"]

activos = st.sidebar.multiselect(
    "Selecciona los activos",
    universo,
    default=["AAPL", "MSFT", "JPM"]
)

periodo = st.sidebar.selectbox(
    "Período de análisis",
    ["3 meses", "6 meses", "1 año"],
    index=2
)

capital = st.sidebar.number_input(
    "Capital inicial (USD)",
    min_value=100,
    value=10000,
    step=500
)

ventana = st.sidebar.slider(
    "Ventana de media móvil",
    5, 30, 10
)

# DATOS SIMULADOS
rng = np.random.default_rng(23)

fechas = pd.bdate_range(
    end=pd.Timestamp.today().normalize(),
    periods=180
)

precios = pd.DataFrame(index=fechas)

for ticker in activos:
    variaciones = rng.normal(
        0.0005, 0.015, len(fechas)
    )

    precios[ticker] = 100 * np.cumprod(
        1 + variaciones
    )

rendimientos = precios.pct_change().dropna()

# CUATRO PESTAÑAS
tab1, tab2, tab3, tab4 = st.tabs([
    "📈 Portafolio",
    "📊 Estadística",
    "⚠️ Riesgo",
    "ℹ️ Proyecto"
])

# PESTAÑA 1
with tab1:

    st.header("Análisis del portafolio")

    c1, c2, c3 = st.columns(3)

    c1.metric("Capital inicial", f"USD {capital:,.0f}")
    c2.metric("Activos seleccionados", len(activos))
    c3.metric("Media móvil", f"{ventana} sesiones")

    if activos:
        base_100 = precios.div(precios.iloc[0]) * 100

        st.subheader("Comparación de activos")
        st.line_chart(base_100)

        st.subheader("Media móvil")
        st.line_chart(precios.rolling(ventana).mean())

        st.caption("Datos simulados para práctica académica.")
    else:
        st.info("Selecciona al menos un activo.")

# PESTAÑA 2
with tab2:

    st.header("Estadística descriptiva")

    if activos:
        st.dataframe(
            rendimientos.describe(),
            use_container_width=True
        )
    else:
        st.info("Selecciona activos para calcular estadísticas.")

# PESTAÑA 3: RIESGO
with tab3:

    st.header("⚠️ Análisis de riesgo financiero")

    if activos:

        volatilidad = rendimientos.std() * np.sqrt(252)

        st.subheader("Volatilidad anualizada")

        st.dataframe(
            (volatilidad * 100).round(2).rename(
                "Volatilidad (%)"
            )
        )

        st.subheader("Evolución del riesgo")

        volatilidad_movil = (
            rendimientos.rolling(20).std() * np.sqrt(252) * 100
        )

        st.line_chart(volatilidad_movil)

        st.info(
            "La volatilidad anualizada mide la variabilidad "
            "de los rendimientos y se calcula multiplicando "
            "la desviación estándar diaria por la raíz "
            "cuadrada de 252."
        )

        # DRAWDOWN DE LA CARTERA EQUIPONDERADA
        cartera = rendimientos.mean(axis=1)

        acumulado = (1 + cartera).cumprod()

        maximos = acumulado.cummax()

        drawdown = acumulado / maximos - 1

        st.subheader("Drawdown del portafolio")

        st.line_chart(drawdown.rename("Drawdown"))

        st.caption(
            "El drawdown representa la caída desde el máximo "
            "anterior del valor acumulado."
        )

    else:
        st.info("Selecciona activos para analizar el riesgo.")

# PESTAÑA 4
with tab4:

    st.header("Acerca de FinanceLab")

    st.write(
        "Aplicación académica desarrollada para estudiar "
        "el comportamiento de activos financieros."
    )

    st.write("Tecnologías utilizadas:")
    st.write("- Python")
    st.write("- Streamlit")
    st.write("- Pandas")
    st.write("- NumPy")
    st.write("- Pillow")

    st.warning(
        "Esta versión utiliza datos simulados. "
        "No representa cotizaciones reales."
    )

st.divider()

st.caption(
    "FinanceLab | Ingeniería Financiera | S23"
)
````

## 07 · Conectar Yahoo Finance y calcular indicadores

Ahora utilizaremos precios históricos de mercado. La app sigue funcionando con datos de ejemplo si Yahoo Finance no responde, **pero muestra claramente cuándo utiliza datos simulados**.

Fórmulas esenciales:

- **Rendimiento diario:** \(r_t = P_t/P_{t-1}-1\).
- **Rendimiento acumulado:** \(R=\prod_t(1+r_t)-1\).
- **Volatilidad anualizada:** \(\sigma_a=\sigma_d\sqrt{252}\).
- **Correlación:** relación lineal entre rendimientos de dos activos, entre −1 y +1.

Para el portafolio de pesos iguales, el rendimiento diario es la media de los rendimientos diarios de los activos. Es un supuesto de **rebalanceo diario** y no incorpora comisiones.

````python

%%writefile app.py

import streamlit as st
import pandas as pd
import numpy as np
import yfinance as yf
import plotly.express as px

st.set_page_config(
    page_title="FinanceLab | Observatorio Financiero",
    page_icon="📈",
    layout="wide"
)

# DISEÑO
st.markdown("""
<style>
.stApp {
    background-color: #0b1425;
    color: #edf4ff;
}
[data-testid="stSidebar"] {
    background-color: #10243a;
}
[data-testid="stMetric"] {
    background-color: #172b43;
    padding: 15px;
    border-radius: 12px;
    border: 1px solid #29415b;
}
h1, h2, h3 {
    color: #5eead4;
}
</style>
""", unsafe_allow_html=True)

# BANNER
st.image(
    "assets/banner_financiero.png",
    use_container_width=True
)

st.title("📈 FinanceLab")
st.subheader("Mi Observatorio Financiero")
st.caption("Ingeniería Financiera | S23")

# CONFIGURACIÓN
st.sidebar.header("⚙️ Configuración")

universo = ["AAPL", "MSFT", "NVDA", "JPM", "GOOGL", "SPY", "GLD"]

activos = st.sidebar.multiselect(
    "Selecciona los activos",
    universo,
    default=["AAPL", "MSFT", "JPM"],
    max_selections=5
)

periodo = st.sidebar.selectbox(
    "Período histórico",
    ["3mo", "6mo", "1y", "2y"],
    index=2
)

capital = st.sidebar.number_input(
    "Capital inicial (USD)",
    min_value=100.0,
    value=10000.0,
    step=500.0
)

fuente = st.sidebar.radio(
    "Fuente de datos",
    ["Yahoo Finance", "Simulados (práctica)"]
)

# DESCARGAR PRECIOS
@st.cache_data(ttl=1800)
def descargar_precios(simbolos, periodo):

    datos = yf.download(
        list(simbolos),
        period=periodo,
        auto_adjust=True,
        progress=False,
        threads=False
    )

    if datos.empty:
        raise ValueError("Yahoo Finance no devolvió datos.")

    if isinstance(datos.columns, pd.MultiIndex):
        cierre = datos["Close"]
    else:
        cierre = datos[["Close"]].rename(
            columns={"Close": simbolos[0]}
        )

    if isinstance(cierre, pd.Series):
        cierre = cierre.to_frame(name=simbolos[0])

    cierre = cierre.reindex(columns=list(simbolos))

    return cierre.sort_index().ffill().dropna()


# DATOS SIMULADOS
def simular_precios(simbolos):

    rng = np.random.default_rng(23)

    fechas = pd.bdate_range(
        end=pd.Timestamp.today().normalize(),
        periods=180
    )

    datos = pd.DataFrame(
        {
            ticker: 100 * np.cumprod(
                1 + rng.normal(0.0005, 0.015, len(fechas))
            )
            for ticker in simbolos
        },
        index=fechas
    )

    return datos


# OBTENER DATOS
if not activos:

    st.info("Selecciona al menos un activo.")
    st.stop()

simulados = fuente == "Simulados (práctica)"

if simulados:

    precios = simular_precios(activos)

else:

    try:

        with st.spinner("Descargando precios históricos..."):

            precios = descargar_precios(
                tuple(activos),
                periodo
            )

        if precios.empty or len(precios) < 30:
            raise ValueError("No hay suficientes datos.")

    except Exception as error:

        st.warning(
            "Yahoo Finance no está disponible. "
            "Se utilizarán datos simulados."
        )

        st.caption(f"Detalle técnico: {error}")

        precios = simular_precios(activos)
        simulados = True


# INDICAR FUENTE
if simulados:

    st.error(
        "MODO DEMOSTRACIÓN: resultados con datos simulados."
    )

else:

    st.success(
        "Datos históricos descargados de Yahoo Finance."
    )


# CÁLCULOS FINANCIEROS
rendimientos = precios.pct_change().dropna()

cartera = rendimientos.mean(axis=1)

acumulado = (1 + cartera).cumprod()

rendimiento_total = acumulado.iloc[-1] - 1

volatilidad = cartera.std(ddof=1) * np.sqrt(252)

valor_inversion = capital * acumulado


# INDICADORES
st.header("📊 Indicadores del portafolio")

c1, c2, c3 = st.columns(3)

c1.metric(
    "Rentabilidad acumulada",
    f"{rendimiento_total:+.2%}"
)

c2.metric(
    "Volatilidad anualizada",
    f"{volatilidad:.2%}"
)

c3.metric(
    "Valor final teórico",
    f"USD {valor_inversion.iloc[-1]:,.2f}"
)

st.caption(
    "Portafolio de pesos iguales con rebalanceo diario. "
    "No incorpora costos ni impuestos."
)


# PESTAÑAS
tab1, tab2, tab3, tab4 = st.tabs([
    "📈 Portafolio",
    "📊 Estadística",
    "⚠️ Riesgo",
    "ℹ️ Proyecto"
])


# PORTAFOLIO
with tab1:

    st.subheader("Comparación de activos")

    base_100 = precios.div(
        precios.iloc[0]
    ) * 100

    st.line_chart(base_100)

    st.subheader("Evolución de la inversión")

    st.line_chart(
        valor_inversion.rename("Valor USD")
    )

    st.info(
        "La base 100 permite comparar activos "
        "con diferentes precios iniciales."
    )


# ESTADÍSTICA
with tab2:

    st.subheader("Estadística descriptiva")

    resumen = pd.DataFrame({
        "Media diaria (%)": rendimientos.mean() * 100,
        "Mediana diaria (%)": rendimientos.median() * 100,
        "Desviación diaria (%)": rendimientos.std() * 100,
        "Mínimo diario (%)": rendimientos.min() * 100,
        "Máximo diario (%)": rendimientos.max() * 100
    })

    st.dataframe(
        resumen.round(3),
        use_container_width=True
    )

    st.subheader("Correlación entre activos")

    fig = px.imshow(
        rendimientos.corr(),
        text_auto=".2f",
        zmin=-1,
        zmax=1,
        color_continuous_scale="RdBu_r"
    )

    st.plotly_chart(
        fig,
        use_container_width=True
    )


# RIESGO
with tab3:

    st.subheader("Volatilidad anualizada")

    volatilidad_activos = (
        rendimientos.std() * np.sqrt(252) * 100
    )

    st.bar_chart(
        volatilidad_activos.rename("Volatilidad (%)")
    )

    st.subheader("Drawdown del portafolio")

    maximos = acumulado.cummax()

    drawdown = acumulado / maximos - 1

    st.line_chart(
        drawdown.rename("Drawdown")
    )

    st.caption(
        "El drawdown mide la caída desde un máximo anterior."
    )


# PROYECTO
with tab4:

    st.subheader("Acerca de FinanceLab")

    st.write(
        "Proyecto académico desarrollado con Python "
        "y Streamlit para estudiar activos financieros."
    )

    st.write("Bibliotecas utilizadas:")

    st.write(
        "Python, Streamlit, Pandas, NumPy, "
        "Plotly y Yahoo Finance."
    )

    st.write(
        "Fuente seleccionada:",
        "Datos simulados" if simulados else "Yahoo Finance"
    )


# DATOS DESCARGABLES
with st.expander("Consultar y descargar datos"):

    st.dataframe(
        precios.tail(30),
        use_container_width=True
    )

    st.download_button(
        "Descargar precios CSV",
        precios.to_csv().encode("utf-8"),
        "precios_financelab.csv",
        "text/csv"
    )

st.divider()

st.caption(
    "FinanceLab | Proyecto académico S23. "
    "Los resultados históricos no garantizan "
    "rentabilidades futuras."
)

````

````python

%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import yfinance as yf

st.set_page_config(
    page_title="FinanceLab | Observatorio Financiero",
    page_icon="📈",
    layout="wide"
)

# DISEÑO
st.markdown("""
<style>
.stApp {background-color:#0b1425;color:#edf4ff;}
[data-testid="stSidebar"] {background-color:#10243a;}
[data-testid="stMetric"] {
    background-color:#172b43;
    padding:15px;
    border-radius:12px;
    border:1px solid #29415b;
}
h1,h2,h3 {color:#5eead4;}
</style>
""", unsafe_allow_html=True)

st.title("📈 FinanceLab")
st.subheader("Mi Observatorio Financiero")
st.caption("Ingeniería Financiera | Laboratorio S23")

# CONFIGURACIÓN
st.sidebar.header("⚙️ Configuración")

universo = ["AAPL", "MSFT", "NVDA", "JPM",
            "GOOGL", "SPY", "GLD"]

activos = st.sidebar.multiselect(
    "Selecciona los activos",
    universo,
    default=["AAPL", "MSFT", "JPM"],
    max_selections=5
)

periodos = {
    "3 meses": "3mo",
    "6 meses": "6mo",
    "1 año": "1y",
    "2 años": "2y"
}

periodo = st.sidebar.selectbox(
    "Período histórico",
    list(periodos.keys()),
    index=2
)

capital = st.sidebar.number_input(
    "Capital inicial (USD)",
    min_value=100.0,
    value=10000.0,
    step=500.0
)

fuente = st.sidebar.radio(
    "Fuente de datos",
    ["Yahoo Finance", "Simulados (práctica)"]
)

# DATOS SIMULADOS
def simular_precios(simbolos):
    rng = np.random.default_rng(23)
    fechas = pd.bdate_range(
        end=pd.Timestamp.today().normalize(),
        periods=180
    )

    datos = pd.DataFrame(index=fechas)

    for ticker in simbolos:
        variaciones = rng.normal(
            0.0005, 0.015, len(fechas)
        )
        datos[ticker] = 100 * np.cumprod(1 + variaciones)

    return datos

# DESCARGA DE DATOS
@st.cache_data(ttl=1800)
def descargar_precios(simbolos, periodo):
    datos = yf.download(
        list(simbolos),
        period=periodo,
        auto_adjust=True,
        progress=False,
        threads=False
    )

    if datos.empty:
        raise ValueError("Yahoo Finance no devolvió datos.")

    if isinstance(datos.columns, pd.MultiIndex):
        cierre = datos["Close"]
    else:
        cierre = datos[["Close"]].rename(
            columns={"Close": simbolos[0]}
        )

    if isinstance(cierre, pd.Series):
        cierre = cierre.to_frame(name=simbolos[0])

    cierre = cierre.reindex(columns=list(simbolos))
    cierre = cierre.sort_index().ffill().dropna()

    if cierre.empty:
        raise ValueError("No existen precios suficientes.")

    return cierre

# OBTENER PRECIOS
if not activos:
    st.warning("Selecciona al menos un activo.")
    st.stop()

simulados = fuente == "Simulados (práctica)"

if simulados:
    precios = simular_precios(activos)
else:
    try:
        with st.spinner("Consultando Yahoo Finance..."):
            precios = descargar_precios(
                tuple(activos),
                periodos[periodo]
            )

        if len(precios) < 2:
            raise ValueError("Datos insuficientes.")

    except Exception as error:
        st.warning(
            "Yahoo Finance no respondió. "
            "Se utilizarán datos simulados."
        )
        st.caption(f"Detalle: {error}")
        precios = simular_precios(activos)
        simulados = True

# FUENTE DE DATOS
if simulados:
    st.error("MODO DEMOSTRACIÓN: datos simulados.")
else:
    st.success("Datos históricos de Yahoo Finance.")

# CÁLCULOS
rendimientos = precios.pct_change().dropna()

if rendimientos.empty:
    st.error("No hay suficientes rendimientos para analizar.")
    st.stop()

cartera = rendimientos.mean(axis=1)
acumulado = (1 + cartera).cumprod()

rentabilidad = acumulado.iloc[-1] - 1
volatilidad = cartera.std(ddof=1) * np.sqrt(252)
valor_final = capital * acumulado.iloc[-1]

# INDICADORES
st.header("📊 Indicadores principales")

c1, c2, c3 = st.columns(3)

c1.metric("Rentabilidad acumulada", f"{rentabilidad:.2%}")
c2.metric("Volatilidad anualizada", f"{volatilidad:.2%}")
c3.metric("Valor final", f"USD {valor_final:,.2f}")

st.caption(
    "Cartera equiponderada con rebalanceo diario. "
    "No incluye comisiones ni impuestos."
)

# PESTAÑAS
tab1, tab2, tab3, tab4, tab5 = st.tabs([
    "📈 Portafolio",
    "📊 Estadística",
    "⚠️ Riesgo",
    "🧠 Interpretación",
    "ℹ️ Proyecto"
])

# PORTAFOLIO
with tab1:
    st.subheader("Comparación de activos: base 100")

    base100 = precios.div(precios.iloc[0]) * 100
    st.line_chart(base100)

    st.subheader("Evolución de la inversión")
    st.line_chart(
        (capital * acumulado).rename("Valor USD")
    )

# ESTADÍSTICA
with tab2:
    st.subheader("Estadística descriptiva")

    resumen = pd.DataFrame({
        "Media diaria (%)": rendimientos.mean() * 100,
        "Mediana diaria (%)": rendimientos.median() * 100,
        "Desviación diaria (%)": rendimientos.std() * 100,
        "Mínimo diario (%)": rendimientos.min() * 100,
        "Máximo diario (%)": rendimientos.max() * 100
    })

    st.dataframe(resumen.round(3), use_container_width=True)

    st.subheader("Correlación entre activos")
    st.dataframe(
        rendimientos.corr().round(3),
        use_container_width=True
    )

# RIESGO
with tab3:
    st.subheader("Volatilidad anualizada por activo")

    vol_activos = rendimientos.std() * np.sqrt(252) * 100
    st.bar_chart(vol_activos.rename("Volatilidad (%)"))

    st.subheader("Drawdown de la cartera")

    maximos = acumulado.cummax()
    drawdown = acumulado / maximos - 1

    st.line_chart(drawdown.rename("Drawdown"))

    st.metric("Máximo drawdown", f"{drawdown.min():.2%}")

# INTERPRETACIÓN
with tab4:
    st.subheader("Interpretación financiera")

    rentabilidad_activos = (
        precios.iloc[-1] / precios.iloc[0] - 1
    ) * 100

    volatilidad_individual = (
        rendimientos.std() * np.sqrt(252) * 100
    )

    resultados = pd.DataFrame({
        "Rentabilidad (%)": rentabilidad_activos,
        "Volatilidad (%)": volatilidad_individual
    })

    st.dataframe(resultados.round(2), use_container_width=True)

    mejor = rentabilidad_activos.idxmax()
    peor = rentabilidad_activos.idxmin()

    st.write(
        f"El activo con mayor rentabilidad fue **{mejor}**, "
        f"con {rentabilidad_activos.max():.2f}%."
    )

    st.write(
        f"El activo con menor rentabilidad fue **{peor}**, "
        f"con {rentabilidad_activos.min():.2f}%."
    )

    st.write(
        f"La volatilidad anualizada de la cartera fue "
        f"**{volatilidad:.2%}**."
    )

    st.write(
        f"El máximo drawdown fue **{drawdown.min():.2%}**."
    )

    if simulados:
        st.warning("Estos resultados utilizan datos simulados.")
    else:
        st.success("Estos resultados utilizan precios históricos.")

# PROYECTO
with tab5:
    st.subheader("Acerca de FinanceLab")

    st.write("""
    Aplicación académica desarrollada para Ingeniería Financiera.

    Tecnologías:
    - Python
    - Streamlit
    - Pandas
    - NumPy
    - Yahoo Finance

    Objetivo: comparar activos, estudiar rendimientos
    y comprender el riesgo de un portafolio.
    """)

# DESCARGA
with st.expander("Consultar y descargar precios"):
    st.dataframe(precios.tail(30), use_container_width=True)

    st.download_button(
        "Descargar precios CSV",
        precios.to_csv().encode("utf-8"),
        "precios_financelab.csv",
        "text/csv"
    )

st.divider()
st.caption(
    "FinanceLab | S23 | Los rendimientos históricos "
    "no garantizan resultados futuros."
)
````
