# El Lápiz Feliz — el menú interactivo
### Clase de ciclos: `while`, `for` y el menú del sistema de facturación

Notebook base para construir el menú en clase, celda por celda. **Cada celda de código de aquí en adelante es un programa completo y ejecutable por sí solo** — no fragmentos sueltos — así que se puede correr en el momento, mostrar el resultado, y seguir agregando en la siguiente celda. La última celda de código es la versión final que se comparte con los estudiantes.

Vamos a practicar cuatro cosas distintas con las opciones del menú:

1. **Registrar una factura** — usa el `for` (repetición *conocida*: el cajero dice cuántos productos hay; como no hemos visto listas, `for` recorre un `range()`, no una colección)
2. **Ver total acumulado del día** — un acumulador que **suma**
3. **Ver cuántas facturas se han registrado** — un acumulador que **cuenta**
4. **Salir** — usa `break` para terminar el `while`

## Paso 1 — el truco del separador

Multiplicar un texto por un número lo repite. Así no hay que escribir `----------------` a mano.

```python
print("-" * 40)
```

## Paso 2 — el menú, solo con `print`

Todavía sin lógica, solo dibujamos cómo se va a ver.

```python
print("-" * 40)
print("EL LÁPIZ FELIZ")
print("-" * 40)
print("1. Registrar una factura")
print("2. Ver total acumulado del día")
print("3. Ver cuántas facturas se han registrado")
print("4. Guardar")
print("-" * 40)
```

## Paso 3 — leer la opción con `input`

Probamos que el programa pueda leer lo que el cajero escribe.

```python
opcion = input("Elige una opción: ")
print("Elegiste la opción:", opcion)
```

## Paso 4 — el `while` que mantiene vivo el menú

Metemos el menú y el `input` dentro de un `while True`. Por ahora, solo la opción `4` hace algo real (`break`); las demás solo confirman que entraron ahí — así probamos que el flujo completo funciona antes de programar la lógica de cada opción.

```python
while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        print("(aquí va a ir registrar la factura)")
    elif opcion == "2":
        print("(aquí va a ir el total del día)")
    elif opcion == "3":
        print("(aquí va a ir el número de facturas)")
    elif opcion == "4":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.selecciona una opcion entre 1-4")
```

## Paso 5 — los acumuladores del día

Necesitamos dos variables que vivan **afuera del `while`**, para que no se borren en cada vuelta del menú:

- `total_dia` — empieza en 0, va a **sumar** el total de cada factura
- `facturas_hechas` — empieza en 0, va a **contar** cuántas facturas se han hecho

Las probamos solas antes de meterlas al programa completo.

```python
total_dia = 0
facturas_hechas = 0

print("total_dia:", total_dia)
print("facturas_hechas:", facturas_hechas)
```

## Paso 6 — opción 1: registrar una factura con `for`

Aquí está el `for` que reemplaza la opción `1`. Como todavía no vimos listas, el cajero dice cuántos productos hay, y el `for` repite exactamente esa cantidad de veces, sumando cada precio.

Dos acumuladores distintos trabajando juntos:
- `total_factura` — se reinicia en 0 **cada vez que se registra una factura** (vive adentro del `if opcion == "1"`)
- `total_dia` y `facturas_hechas` — **no se reinician**, siguen creciendo factura tras factura, porque viven afuera del `while`

De aquí en adelante, cada celda repite el programa completo — lo nuevo está marcado con `# 👈 NUEVO`.

```python
total_dia = 0
facturas_hechas = 0

while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        cantidad = int(input("¿Cuántos productos tiene esta factura? "))  # 👈 NUEVO
        total_factura = 0  # 👈 NUEVO

        for i in range(cantidad):  # 👈 NUEVO
            print(f"Producto {i + 1} de {cantidad}:")
            nombre = input("  Nombre del producto: ")
            precio = float(input("  Precio: "))
            total_factura = total_factura + precio

        print(f"Factura registrada. Total: ${total_factura:,.0f}")  # 👈 NUEVO
        total_dia = total_dia + total_factura  # 👈 NUEVO
        facturas_hechas = facturas_hechas + 1  # 👈 NUEVO
    elif opcion == "2":
        print("(aquí va a ir el total del día)")
    elif opcion == "3":
        print("(aquí va a ir el número de facturas)")
    elif opcion == "4":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.")
```

## Paso 7 — opciones 2 y 3: mostrar lo acumulado

Solo falta mostrar `total_dia` y `facturas_hechas`. Esta es la versión final y completa del menú.

```python
total_dia = 0
facturas_hechas = 0

while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        cantidad = int(input("¿Cuántos productos tiene esta factura? "))
        total_factura = 0

        for i in range(cantidad):
            print(f"Producto {i + 1} de {cantidad}:")
            nombre = input("  Nombre del producto: ")
            precio = float(input("  Precio: "))
            total_factura = total_factura + precio

        print(f"Factura registrada. Total: ${total_factura:,.0f}")
        total_dia = total_dia + total_factura
        facturas_hechas = facturas_hechas + 1
    elif opcion == "2":
        print(f"Total acumulado del día: ${total_dia:,.0f}")  # 👈 NUEVO
    elif opcion == "3":
        print(f"Facturas registradas hoy: {facturas_hechas}")  # 👈 NUEVO
    elif opcion == "4":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.")
```
